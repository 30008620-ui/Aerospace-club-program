```python
import matplotlib.pyplot as plt

# telementry log
# time in min, altitude in meters
time = [0, 4, 9, 15, 21, 26, 30]
alt = [0, 1200, 3100, 3100, 3000, 900, 0]

# find max height
highest_alt = 0
time_at_max = 0

for i in range(len(alt)):
  if alt[i] > highest_alt:
     highest_alt = alt[i]
     time_at_max = time[i]

   # average climb speed
climb_rate = highest_alt / time_at_max

print("max altitude:" , highest_alt)
print("time at max:" , time_at_max)
print("climb rate:" , climb_rate)

# plot graph
plt.plot(time, alt)
plt.xlabel("Time (min)")
plt.ylabel("Altitude (m)")
plt.title("Airplane Flight Telemetry")
plt.show()
```
